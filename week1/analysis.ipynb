```python
g = (req.groupby("channel", observed=True)
        .agg(n=("request_id", "size"),
             rated=("citizen_rating", "count"),
             mean_rating=("citizen_rating", "mean")))
g
g["response_rate"] = (g["rated"] / g["n"] * 100).round(1)
g.sort_values("n", ascending=False)
```

```python
MAP = {"Waste Collection": "waste_collection",
       "street light": "streetlight", "Streetlight": "streetlight"}

c = req[req["status"].eq("closed")].copy()
c["hours"] = (c["closed_at"] - c["created_at"]).dt.total_seconds() / 3600
c = c[c["hours"] >= 0].copy()
c["cat"] = c["category"].replace(MAP)
c["missed"] = c["hours"] > c["target_hours"]

q3 = c.groupby(["cat", "priority"], observed=True).agg(
        n=("request_id", "size"),
        med_h=("hours", "median"),
        miss_rate=("missed", "mean"))
q3 = q3[q3["n"] >= 100]
q3 = q3.sort_values("miss_rate", ascending=False).head(10)
q3["miss_rate"] = (q3["miss_rate"] * 100).round(1)
q3
```

```python
dist = pd.read_csv(RAW/"districts.csv")
cat = pd.read_csv(RAW/"bi_dim_category.csv")
print(dist.shape, cat.shape)

j = req.merge(dist,on="district_id", how="left", validate="m:1")
print(len(req), len(j),j["district_name"].isna().sum())
```

```python
t_inner = req.merge(cat, left_on="category", right_on="cat", how="inner")
print(len(t_inner))

t_left = req.merge(cat,left_on="category", right_on="cat",
                    how="left",indicator=True)
print(t_left["_merge"].value_counts())
print(t_left[t_left["_merge"] == "left_only"]["category"].value_counts())
```

```python
budget = pd.DataFrame({
    "department": ["Roads & Lighting"] * 2 + ["Water"] * 2,
    "year":       [2024, 2025, 2024, 2025],
    "budget":     [1.2e6, 1.25e6, 4.0e5, 4.2e5],
})
bad = req.merge(budget, on="department", how="left")
print("bad:", len(bad))                             # 44246
print(bad["year"].value_counts(dropna=False).to_dict())
# {nan: 19356, 2024.0: 12445, 2025.0: 12445}
```

```python
try:
    req.merge(budget, on="department", validate="m:1")
except Exception as e:
    print(type(e).__name__, ":", e)
```

行数变化三原因：键重复 → 爆炸；键对不上 → 丢行/NaN；基表口径变 → 整体位移。
validate 防第一种，indicator 查第二种，第三种靠 7 个 KPI 对账。
